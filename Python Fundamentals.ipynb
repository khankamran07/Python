### 1. Identifiers/Variables in Python?

**Identifiers are the names we give to different things in Python**, such as variables, functions, classes, etc.

In simple words:

> **Identifier = Name given to something in a Python program.**

### Example

```python
name = "Kamran"
age = 22
```

Here:

- `name` → identifier
- `age` → identifier

We give these names so that we can easily use the values later.

```python
name = "Kamran"
print(name)
```

Output:

```text
Kamran
```

### Where can we use identifiers?

Identifiers can be used for:

```python
student_name = "Rahul"     # variable
```

```python
def calculate_salary():    # function
    pass
```

```python
class Student:             # class
    pass
```

So, **`student_name`, `calculate_salary`, and `Student` are all identifiers.**

### 📌 Rules for Python Identifiers

1. **Can contain letters** → `name`, `student`
2. **Can contain numbers** → `student1`
3. **Can use underscore `_`** → `student_name`
4. **Cannot start with a number** ❌
   ```python
   1student = "Rahul"  # Wrong
   ```
5. **Cannot contain spaces** ❌
   ```python
   student name = "Rahul"  # Wrong
   ```
6. **Cannot use Python keywords** ❌
   ```python
   class = 10  # Wrong
   ```

#### 1.1 String Indexing

- The process of obtaining particular character from given string by passing valid index is called string indexing.
- Syntax :-  strobj[index]
- Here index can be either +ve/-ve, valid index.
-  If the index value is invalid, we get IndexError.
-  Consider the following statement and its memory management diagram.

````python
s = "PYTHON"
````

````python
# Indexing
# Ex

print(s[0])
print(s[-6])
print(s[-3])
print(s[4])
print(s[1:4])
print(s[40]) # IndexError                                
````

#### 1.2 String Slicing
- The process obtaining range of character/sub-string from the given main string is called string slicing.  
   Syntax :- strobject[begin:end]
- This center give range of characters from begin index to end-1 index provided begin index < end index, otherwise we never get any output(' ').

````python
# Slicing 
# Ex

print(s[1:5])
print(s[0:6:2])
print(s[::])
print(s[2:100])
print(s[-100:-2])
print(s[2:4])
print(s[0:4])
print(s[12:1])
print(s[-6:-12])
````

#### Special points :-
- If we don't specify end index, then Python environment takes no of char-1 as end index position.
- If we don't specify begin index, then the Python environment takes zero(initial position) as begin index/position.
- If we don't specify both begin and end index, then zero(initial position) as begin index/position and no of character-1 as end index position.

````python
z = "PYTHON"
print(z[1:50])
print(z[-100:-2])
print(z[1:])
print(z[:4])
print(z[:])
````

### Range data type

- Range is one of the predefined class.
- The purpose of the data type is to store sequence of value by maintaining equal interval difference.
- An object of range is immutable.

**Syntax 1** (range start)  
This syntax generates a range of values from 0 to start -1 values.


````python
# Exampe 

r = range(6)
print(type(r))

for x in r:
    print(x) # default step value is 1
````

**Syntax 2** range(start,stop)    
This syntax generates a range of values from start to stop -1 values.

````python
# Example --> generate value from 11 to 20

for x in range(11,21):
    print(x)
````

````python
# Example --> generate value from 1000 to 1005

for x in range(1000,1006):
    print(x)
````

**Syntax 3** range(start,stop,step)    
This syntax generates a range of values from start to stop -1 with step values and can be +ve or -ve .

````python
# Example --> generate even no b/w 30 to 50

for x in range(30,51,2):
    print(x)
````

````python
# Example --> generate odd no b/w 50 and 40

for x in range(49,40,-2):
    print(x)
````

````python
# Example --> generate -1 to -10
for x in range (-1,-11,-1):
    print(x)
````

````python
# Example --> generate -50, -40, -30, -20, -10
for x in range (-50,-9,10):
    print(x)
````

**Note :** On the object of range we can apply indexing & slicing operations.

````python
# example - Consider the following 

r = range(10,16)

# for the above statement, the memory management diagram is required below

print(r[0])
print(r[-6])
print(r[4])
````

````python
for x in r[2:5]:
    print(x)
````

### 6. copy()
- This function is used to copy the contents of source list object into destination list object with same content with different address. This copy    type of copy process is called shallow copy.  
  **Syntax: dest.listObj = sourcelistObj.copy()**

````python
l1 = [10,20,"hyd"]
l2 = l1.copy() #shallow copy
print(l1,l2)
print(id(l1),id(l2))
````

**Points of shallow copy**
- Initial content of source and destination list object are same.
- Address of source and destination list object are different.
- Modification are reflected to each other dependent.

**Points of deep copy**
- Initial content of source and destination list object are same.
- Address of source and destination list object are same.
- Modifications are reflected to each other dependent.  
  **Syntax: destlistobj = sourcelistobj.**

````python
# example 

l1 = [10,'Python']
l2 = l1 # Deep copy
print(l1,l2)
print(id(l1),id(l2))
l1.append('AI')
print(l1,l2)
````

**Special Points(Another way of shallow copy with slicing)**

````python
# example

l1 = [10,'hyd']
l2 = l1[:] #shallow copy with slicing
print(l1,l2)
print(id(l1),id(l2))
l1.append('india')
print(l1,l2)
````

### v) dict category data types
- This data type allows us to store the data in the form of (key-value) pair.
- In (key-value), the value of key represent the unique and the value represent may or may not unique.
- The elements dict object must be enclosed within curly braces{} and whose key-value must be separated by (:) and terminated by (;).
- To convert one type of value into dict type, we use dict(), and it can also use to create empty dict object.
- An object of dict is immutable.

#### create a dict:
- we create 2 types of dict objects. They are  
  a) empty dict  
  b) non-empty dict  

**syntax for empty dict**  
dictobj = {}  
or  
dictobj=dict{}

````python
d1 = {}
# or 
d1 = dict()
print(d1,type(d1))
````

**syntax: for adding(key & value) to empty dict object**
- dictobj[keyname - 1] = val-1
- dictobj[keyname - 2] = val-2
--------------------------
- dictobj[keyname - n] = val-n

**Note:** Here if keyname is str type, then keyname must written in single/double quotes. If keyname is numeric, we write keyname directly without quotes.

````python
# example

d1['stno'] = 123
d1[12] = 'OUCET'
print(d1)
````

````python
d1['place']='hyd'
print(d1)
````

**syntax for creating non-empty dicts**  
dictobj = {keyname-1:val1,keyname-2:val2,...keyname-n:valn}

````python
# example

d1 = {'stno':100,12:'oucet','place':'hyd'}
print(d1)
````

**updating the value of dict**  
**syntax:** dictobj[keyname] = new value
- If key name is already existing in dict object, then old value is replaced with new value. If keyname is not existing in dict object, then (keyname,new value) is inserted newly in dict object.


````python
# example 

d1 = {10:'AI',11:'Python'}
d1[10] = 'DS'
print(d1)
d1[12] = 'web dev'
print(d1)
````
